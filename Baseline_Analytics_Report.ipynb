## 12. Key Business Insights

### 1. Early-tenure customers are high risk

Customers in their first 12 months have a churn rate of **47.44%**, compared with only **9.51%** among customers with 49–72 months of tenure.

### 2. Contract commitment is strongly associated with churn

Month-to-month customers have a churn rate of **42.71%**, while two-year customers have a churn rate of only **2.83%**.

### 3. The highest-risk segment is clearly identifiable

Month-to-month customers with 0–12 months tenure have the highest observed churn rate at **51.35%**.

### 4. Payment method is an important signal

Electronic-check customers show a churn rate of **45.29%**.

### 5. Service configuration matters

Fiber optic customers have a churn rate of **41.89%**, while customers without internet service have a churn rate of **7.40%**.

### 6. Support and security services show strong differences

Customers without Online Security or Tech Support have churn rates above **41%**, while customers using these services have considerably lower observed churn.

## 13. Business Recommendations

Based on the baseline analysis, the following strategies can be considered:

### Customer Onboarding

Strengthen onboarding and engagement during the first 12 months.

### Contract Retention

Develop targeted incentives to encourage high-risk month-to-month customers to move toward longer-term contracts.

### High-Risk Customer Monitoring

Prioritise customers with:

- Short tenure
- Month-to-month contracts
- Electronic check payment
- Fiber optic service
- No Online Security
- No Tech Support

### Service Engagement

Investigate whether promoting relevant support and security services can improve customer retention.

### Payment Method Investigation

Investigate why electronic-check customers show substantially higher churn and evaluate whether payment experience contributes to customer dissatisfaction.

## 14. Baseline Analytics Conclusion

The exploratory analysis identified several customer characteristics associated with higher observed churn.

The strongest baseline patterns were associated with **contract type, customer tenure, payment method, internet service and support/security services**.

The highest observed churn segment was **month-to-month customers with 0–12 months tenure**, with a churn rate of **51.35%**.

These findings provide a baseline understanding of customer churn and will guide the next stage of the project:

```python
import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt

# Load dataset
df = pd.read_csv("WA_Fn-UseC_-Telco-Customer-Churn.csv")

print("Day 7 EDA environment is working!")
print("Dataset shape:", df.shape)
```

```python
contract_rate = (
    df.groupby("Contract")["Churn"]
    .apply(lambda x: (x == "Yes").mean() * 100)
    .reset_index(name="ChurnRate")
)

print(contract_rate)
```

```python
plt.figure(figsize=(10, 6))

ax = sns.barplot(
    data=contract_rate,
    x="Contract",
    y="ChurnRate",
    errorbar=None
)

for container in ax.containers:
    ax.bar_label(
        container,
        fmt="%.1f%%",
        padding=5
    )

plt.title(
    "Customer Churn Rate by Contract Type",
    fontsize=18,
    fontweight="bold"
)

plt.xlabel("Contract Type")
plt.ylabel("Churn Rate (%)")

plt.ylim(0, 50)
plt.grid(axis="y", linestyle="--", alpha=0.3)

plt.tight_layout()
plt.show()
```

```python
contract_rate = (
    df.groupby("Contract")["Churn"]
    .apply(lambda x: (x == "Yes").mean() * 100)
    .reset_index(name="ChurnRate")
)

plt.figure(figsize=(10, 6))

ax = sns.barplot(
    data=contract_rate,
    x="Contract",
    y="ChurnRate",
    errorbar=None
)

for container in ax.containers:
    ax.bar_label(
        container,
        fmt="%.1f%%",
        padding=5,
        fontsize=11,
        fontweight="bold"
    )

plt.title(
    "Customer Churn Rate by Contract Type",
    fontsize=18,
    fontweight="bold",
    pad=20
)

plt.suptitle(
    "Longer contract commitments are associated with lower observed churn",
    fontsize=11,
    y=0.93
)

plt.xlabel("Contract Type", fontsize=12, fontweight="bold")
plt.ylabel("Churn Rate (%)", fontsize=12, fontweight="bold")

plt.ylim(0, 50)
plt.grid(axis="y", linestyle="--", alpha=0.3)

plt.tight_layout()
plt.show()
```

```python
df["TenureGroup"] = pd.cut(
    df["tenure"],
    bins=[-1, 12, 24, 48, 72],
    labels=[
        "0–12 Months",
        "13–24 Months",
        "25–48 Months",
        "49–72 Months"
    ]
)

tenure_rate = (
    df.groupby("TenureGroup", observed=True)["Churn"]
    .apply(lambda x: (x == "Yes").mean() * 100)
    .reset_index(name="ChurnRate")
)

plt.figure(figsize=(11, 6))

ax = sns.barplot(
    data=tenure_rate,
    x="TenureGroup",
    y="ChurnRate",
    errorbar=None
)

for container in ax.containers:
    ax.bar_label(
        container,
        fmt="%.1f%%",
        padding=5,
        fontsize=11,
        fontweight="bold"
    )

plt.title(
    "Customer Churn Rate by Tenure",
    fontsize=18,
    fontweight="bold",
    pad=20
)

plt.suptitle(
    "Early-tenure customers show substantially higher churn risk",
    fontsize=11,
    y=0.93
)

plt.xlabel("Customer Tenure", fontsize=12, fontweight="bold")
plt.ylabel("Churn Rate (%)", fontsize=12, fontweight="bold")

plt.ylim(0, 55)
plt.grid(axis="y", linestyle="--", alpha=0.3)

plt.tight_layout()
plt.show()
```

```python
payment_churn = (
    df.groupby("PaymentMethod")["Churn"]
    .apply(lambda x: (x == "Yes").mean() * 100)
    .reset_index(name="ChurnRate")
    .sort_values("ChurnRate", ascending=True)
)

plt.figure(figsize=(11, 6))

ax = sns.barplot(
    data=payment_churn,
    x="ChurnRate",
    y="PaymentMethod",
    errorbar=None
)

for container in ax.containers:
    ax.bar_label(
        container,
        fmt="%.1f%%",
        padding=5,
        fontsize=10,
        fontweight="bold"
    )

plt.title(
    "Customer Churn Rate by Payment Method",
    fontsize=18,
    fontweight="bold",
    pad=20
)

plt.suptitle(
    "Electronic check customers show the highest observed churn",
    fontsize=11,
    y=0.93
)

plt.xlabel("Churn Rate (%)", fontsize=12, fontweight="bold")
plt.ylabel("Payment Method", fontsize=12, fontweight="bold")

plt.xlim(0, 50)
plt.grid(axis="x", linestyle="--", alpha=0.3)

plt.tight_layout()
plt.show()
```

```python
internet_churn = (
    df.groupby("InternetService")["Churn"]
    .apply(lambda x: (x == "Yes").mean() * 100)
    .reset_index(name="ChurnRate")
    .sort_values("ChurnRate", ascending=False)
)

plt.figure(figsize=(10, 6))

ax = sns.barplot(
    data=internet_churn,
    x="InternetService",
    y="ChurnRate",
    errorbar=None
)

for container in ax.containers:
    ax.bar_label(
        container,
        fmt="%.1f%%",
        padding=5,
        fontsize=11,
        fontweight="bold"
    )

plt.title(
    "Customer Churn Rate by Internet Service",
    fontsize=18,
    fontweight="bold",
    pad=20
)

plt.suptitle(
    "Fiber optic customers show the highest observed churn",
    fontsize=11,
    y=0.93
)

plt.xlabel("Internet Service", fontsize=12, fontweight="bold")
plt.ylabel("Churn Rate (%)", fontsize=12, fontweight="bold")

plt.ylim(0, 50)
plt.grid(axis="y", linestyle="--", alpha=0.3)

plt.tight_layout()
plt.show()
```

```python
contract_tenure = (
    df.groupby(
        ["Contract", "TenureGroup"],
        observed=True
    )["Churn"]
    .apply(lambda x: (x == "Yes").mean() * 100)
    .reset_index(name="ChurnRate")
)

heatmap_data = contract_tenure.pivot(
    index="Contract",
    columns="TenureGroup",
    values="ChurnRate"
)

plt.figure(figsize=(12, 7))

ax = sns.heatmap(
    heatmap_data,
    annot=True,
    fmt=".1f",
    linewidths=2,
    linecolor="white",
    cbar_kws={
        "label": "Churn Rate (%)",
        "shrink": 0.8
    },
    annot_kws={
        "fontsize": 13,
        "fontweight": "bold"
    }
)

plt.title(
    "CUSTOMER CHURN RISK MATRIX",
    fontsize=20,
    fontweight="bold",
    pad=25
)

plt.suptitle(
    "Contract commitment and tenure jointly identify high-risk customer segments",
    fontsize=11,
    y=0.94
)

plt.xlabel(
    "Customer Tenure",
    fontsize=13,
    fontweight="bold",
    labelpad=12
)

plt.ylabel(
    "Contract Type",
    fontsize=13,
    fontweight="bold",
    labelpad=12
)

plt.xticks(fontsize=11, rotation=0)
plt.yticks(fontsize=11, rotation=0)

plt.tight_layout()
plt.show()
```
