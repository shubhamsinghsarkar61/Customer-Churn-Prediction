## Single Customer LTV Inference API

Adds a validated single-customer prediction endpoint
to the Customer LTV FastAPI service

```python
from fastapi import FastAPI

app = FastAPI(
    title="Customer LTV Prediction API",
    description="FastAPI service for single-customer LTV inference.",
    version="1.0.0"
)

@app.get("/")
def read_root():
    return {
        "message": "Customer LTV Prediction API",
        "status": "running",
        "version": "1.0.0"
    }

@app.get("/health")
def health_check():
    return {
        "status": "healthy",
        "service": "Customer LTV Prediction API"
    }
```

```python
from fastapi import FastAPI
```

```python
class CustomerInput(BaseModel):
    tenure_months: float = Field(..., ge=0)
    monthly_charges: float = Field(..., ge=0)
    total_charges: float = Field(..., ge=0)
```

```python
from pydantic import BaseModel, Field
```

```python
@app.post("/predict")
def predict_ltv(customer: CustomerInput):
    features = {
        "tenure_months": customer.tenure_months,
        "monthly_charges": customer.monthly_charges,
        "total_charges": customer.total_charges
    }

    return {
        "status": "success",
        "message": "Customer data validated for LTV inference",
        "features": features
    }
```

```python
customer = CustomerInput(
    tenure_months=24,
    monthly_charges=70.0,
    total_charges=1680.0
)

print(customer)
```

```python
result = predict_ltv(customer)
print(result)
```

```python
# Test invalid customer input

try:
    invalid_customer = CustomerInput(
        tenure_months=-5,
        monthly_charges=70.0,
        total_charges=1680.0
    )
    print(invalid_customer)

except Exception as e:
    print("Validation Error:", e)
```

```python
# Test another valid customer

customer_2 = CustomerInput(
    tenure_months=48,
    monthly_charges=85.0,
    total_charges=4080.0
)

result_2 = predict_ltv(customer_2)

print(result_2)
```
